```python
import cv2
from ultralytics import YOLO

# Load your custom YOLOv8 model
model = YOLO('train/weights/best.pt')

# Open webcam (0 = default camera)
cap = cv2.VideoCapture(0)

# Check if camera opened successfully
if not cap.isOpened():
    print("Error: Could not open webcam.")
    exit()

# Class names from the model
class_names = model.names

while True:
    # Read a frame from the webcam
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    # Run YOLOv8 inference on the frame
    results = model(frame)[0]

    # Draw results
    for box in results.boxes:
        # Get coordinates and class
        x1, y1, x2, y2 = map(int, box.xyxy[0])  # Bounding box
        cls_id = int(box.cls[0])               # Class ID
        conf = float(box.conf[0])              # Confidence score
        label = f"{class_names[cls_id]} {conf:.2f}"

        # Draw rectangle and label
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(frame, label, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)

    # Show the frame
    cv2.imshow("YOLOv8 Live Detection", frame)

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Clean up
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
from ultralytics import YOLO

# Load YOLOv8 model
model = YOLO('train/weights/best.pt')

# Set minimum confidence threshold
CONF_THRESHOLD = 0.4

# Path to the input image
image_path = 'C:/Users/SAMRUDHI/OneDrive/Desktop/Final year project/GarbageDetection/YOLOV8/Code/simple_images_original_cardboard-in-the-trash_40-jpeg_359a880e-1718-46ab-aeb1-0e6471280e8b_jpeg.rf.36cc288cd3459a94406850e86e62b1fd.jpg'  # ← change this

# Read the image
image = cv2.imread(image_path)

# Run inference
results = model(image)[0]

# Class names
class_names = model.names

# Loop through detected objects
for box in results.boxes:
    conf = float(box.conf[0])
    cls_id = int(box.cls[0])

    if conf >= CONF_THRESHOLD:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        label = f"{class_names[cls_id]} {conf:.2f}"

        # Draw bounding box and label
        cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(image, label, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)

        # Print detections
        print(f"Detected: {class_names[cls_id]}, Confidence: {conf:.2f}")

# Show image with detections
cv2.imshow("Detected Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
